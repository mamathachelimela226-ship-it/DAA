```python
import pandas as pd
import numpy as np

np.random.seed(78)

df = pd.DataFrame({
'Date': pd.date_range('2024-01-01', periods=500),
'Category': np.random.choice(['Food','Transport','Rent','Shopping','Bills','Health','Entertainment'], 500),
'Amount': np.random.randint(100, 8000, 500),
'Payment_Mode': np.random.choice(['UPI','Cash','Card'], 500),
'Location': np.random.choice(['Hyderabad','Online','Gachibowli'], 500),
'Month': np.random.choice(['Jan','Feb','Mar','Apr','May','Jun','Jul','Aug','Sep','Oct','Nov','Dec'], 500)
})

df.to_csv('expense_dataset_78.csv', index=False)
df.head()
```

```python
import matplotlib.pyplot as plt

# Graph 1
df.groupby('Category')['Amount'].sum().plot(kind='bar', color='orange')
plt.title('78 - Category Spending')
plt.show()

# Graph 2
df['Payment_Mode'].value_counts().plot(kind='pie', autopct='%1.1f%%')
plt.title('Payment Mode')
plt.show()

# Graph 3
df.groupby('Month')['Amount'].sum().plot(marker='o')
plt.title('Monthly Trend')
plt.show()
```

```python
import pandas as pd
import numpy as np

# Roll No 78
np.random.seed(78)

# 500 proper dates - NO TIME
dates = pd.date_range(start='2025-01-01', end='2025-06-30', periods=500)
dates = dates.normalize().strftime('%d-%m-%Y')  # DD-MM-YYYY

# Data
df = pd.DataFrame({
    'Date': dates,
    'Category': np.random.choice(['Food','Transport','Bills','Shopping','Entertainment','Health','Education'], 500),
    'Amount': np.random.randint(200, 8000, 500),
    'Payment_Mode': np.random.choice(['Cash','UPI','Card','Netbanking'], 500),
    'Location': np.random.choice(['Hyderabad','Secunderabad','Kukatpally','Banjara Hills','Ameerpet'], 500)
})

# Shuffle
df = df.sample(frac=1, random_state=78).reset_index(drop=True)
df = df.sort_values(by='Date', key=lambda x: pd.to_datetime(x, format='%d-%m-%Y'))

# Save & Download
df.to_csv('expense_dataset_78_FINAL.csv', index=False)

print("✅ FINAL FILE READY - 500 rows")
print(df.head())
print("\nDate Format:", df['Date'].iloc[0], "- DD-MM-YYYY")

from google.colab import files
files.download('expense_dataset_78_FINAL.csv')
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# 1. FINAL 2025 file load chey
df = pd.read_csv('expense_dataset_78_FINAL.csv')

# Date correct format lo marchu
df['Date'] = pd.to_datetime(df['Date'], format='%d-%m-%Y')
df['Month'] = df['Date'].dt.month_name()

print("2025 Data Loaded - Rows:", len(df))
df.head()

# 2. GRAPH 1: Monthly Spending
plt.figure(figsize=(10,5))
monthly = df.groupby('Month')['Amount'].sum()
plt.plot(monthly.index, monthly.values, marker='o')
plt.title('Monthly Expense - 2025')
plt.ylabel('Amount')
plt.show()

# 3. GRAPH 2: Category Wise
plt.figure(figsize=(8,6))
df.groupby('Category')['Amount'].sum().plot(kind='bar')
plt.title('Category Wise Spending - 2025')
plt.show()

# 4. GRAPH 3: Payment Mode Pie
plt.figure(figsize=(6,6))
df.groupby('Payment_Mode')['Amount'].sum().plot(kind='pie', autopct='%1.1f%%')
plt.title('Payment Mode - 2025')
plt.show()
```

```python
# Fix Month Order
month_order = ['January','February','March','April','May','June','July','August','September','October','November','December']

# 1. Monthly Spending - NEAT LINE
plt.figure(figsize=(12,5))
monthly = df.groupby(df['Date'].dt.month)['Amount'].sum()
monthly.index = [month_order[i-1] for i in monthly.index]
plt.plot(monthly.index, monthly.values, marker='o', linewidth=2)
plt.title('Monthly Expense Trend - 2025 (Recent Year)', fontsize=14, fontweight='bold')
plt.xticks(rotation=45)
plt.grid(True)
plt.show()

# 2. Category Wise - NEAT BAR
plt.figure(figsize=(10,5))
cat = df.groupby('Category')['Amount'].sum().sort_values(ascending=False)
plt.bar(cat.index, cat.values, color='skyblue')
plt.title('Category Wise Spending - 2025', fontsize=14, fontweight='bold')
plt.xticks(rotation=30)
plt.show()

# 3. Payment Mode - PIE
plt.figure(figsize=(7,7))
pay = df.groupby('Payment_Mode')['Amount'].sum()
plt.pie(pay.values, labels=pay.index, autopct='%1.1f%%', startangle=90)
plt.title('Payment Mode Distribution - 2025', fontsize=14, fontweight='bold')
plt.show()

print("✅ All 3 Graphs Ready for 2025 FINAL Data!")
```
