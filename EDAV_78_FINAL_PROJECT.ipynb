```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

np.random.seed(78)
dates = pd.date_range('2025-01-01', '2025-06-30', periods=500)
categories = ['Food','Transport','Bills','Shopping','Entertainment','Health']
payment = ['Cash','UPI','Card','Netbanking']
locations = ['Hyderabad','Secunderabad','Kukatpally','Banjara Hills']

df = pd.DataFrame({
    'Date': dates,
    'Category': np.random.choice(categories, 500),
    'Amount': np.random.randint(50, 5000, 500),
    'Payment_Mode': np.random.choice(payment, 500),
    'Location': np.random.choice(locations, 500)
})

print("Rows:", len(df))
df.head()

plt.figure(figsize=(10,4))
df.groupby(df['Date'].dt.month)['Amount'].sum().plot(marker='o')
plt.title('Monthly Expense - 2025')
plt.show()

plt.figure(figsize=(8,4))
df.groupby('Category')['Amount'].sum().plot(kind='bar', color='skyblue')
plt.title('Category Wise')
plt.show()

plt.figure(figsize=(6,6))
df['Payment_Mode'].value_counts().plot(kind='pie', autopct='%1.1f%%')
plt.title('Payment Mode')
plt.show()
```
